# Data analysis with pandas

```python
import pandas as pd
```

## Load data from CSV file into Pandas dataframe

```python
countries_df = pd.read_csv('countries.csv')
```

```python
countries_df
```

### Information of dataframe

```python
countries_df.info()
```

## How many countries does the dataframe contain ?

```python
num_countries = countries_df.shape[0]
```

```python
print('There are {} countries in the dataset'.format(num_countries))
```

## Retrieve a list of continents from the dataframe ?

```python
continents = countries_df.continent.unique()
```

```python
continents
```

## What is the total population of all the countries listed in this dataset ?

```python
total_population = countries_df.population.sum()
```

```python
print('The total population is {}.'.format(int(total_population)))
```

## What is the overall life expectancy across in the world ?

```python
overall_life_expectancy = ((countries_df.life_expectancy * countries_df.population).sum() / total_population.sum())
```

```python
overall_life_expectancy
```

## Create a dataframe containing 10 countries with the highest population ?

```python
most_populous_df = countries_df.sort_values('population', ascending=False).head(10)
```

```python
most_populous_df
```

## Add a new column in countries_df to record the overall GDP per country (product of population & per capita GDP)

```python
countries_df['gdp'] = (countries_df.population * countries_df.gdp_per_capita)
```

```python
countries_df
```

## Create a dataframe containing 10 countries with the lowest GDP per capita, among the counties with population greater than 100 million

```python
required_df = countries_df[countries_df.population > 1e8].sort_values('gdp_per_capita').head(10)
```

```python
required_df
```

## Create a data frame that counts the number countries in each continent ?

```python
country_counts_df = countries_df.groupby('continent').count()
```

```python
country_counts_df
```

## Create a data frame showing the total population of each continent ?

```python
countries_df = countries_df.drop(columns=['location'])
```

```python
continent_populations_df = countries_df.groupby('continent').sum()
```

```python
continent_populations_df
```
